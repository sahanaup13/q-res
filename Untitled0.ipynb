```python
loads = [
    # (name, power demand in kW, priority 1-10)
    ("Hospital",      40, 10),
    ("Cell Tower",    20,  8),
    ("Water Pump",    25,  7),
    ("Shelter",       15,  6),
    ("Street Lights", 10,  2),
]
battery_capacity = 60  # kW available

for name, demand, priority in loads:
    print(f"{name:14} demand={demand} kW  priority={priority}")
print("Battery capacity:", battery_capacity, "kW")
```

```python
from itertools import product

best_choice = None
best_priority = -1

# Try every ON/OFF combination: 2^5 = 32 possibilities
for choice in product([0, 1], repeat=len(loads)):
    total_demand = sum(c * l[1] for c, l in zip(choice, loads))
    total_priority = sum(c * l[2] for c, l in zip(choice, loads))

    if total_demand <= battery_capacity and total_priority > best_priority:
        best_priority = total_priority
        best_choice = choice

print("Best choice:", best_choice)
print("Powered loads:", [l[0] for c, l in zip(best_choice, loads) if c == 1])
print("Total priority:", best_priority)
print("Power used:", sum(c * l[1] for c, l in zip(best_choice, loads)), "kW")
```

```python
from qiskit_optimization import QuadraticProgram

qp = QuadraticProgram("Q-RESTORE")

# One binary variable per load: x0..x4 (1 = powered, 0 = off)
for i in range(len(loads)):
    qp.binary_var(name=f"x{i}")

# Goal: maximize total priority
qp.maximize(linear={f"x{i}": loads[i][2] for i in range(len(loads))})

# Rule: total demand must not exceed the battery
qp.linear_constraint(
    linear={f"x{i}": loads[i][1] for i in range(len(loads))},
    sense="<=",
    rhs=battery_capacity,
    name="battery",
)

print(qp.prettyprint())
```

```python
from qiskit_algorithms import QAOA
from qiskit_algorithms.optimizers import COBYLA
from qiskit_aer.primitives import Sampler
from qiskit_optimization.algorithms import MinimumEigenOptimizer

# QAOA = the quantum algorithm, COBYLA = the classical tuner it uses
qaoa = QAOA(sampler=Sampler(), optimizer=COBYLA(maxiter=100), reps=2)

solver = MinimumEigenOptimizer(qaoa)
result = solver.solve(qp)

print("QAOA choice:", [int(v) for v in result.x])
print("Powered loads:", [l[0] for v, l in zip(result.x, loads) if v == 1])
print("Total priority:", result.fval)
```

```python
import time
from itertools import product

def brute_force(loads, cap):
    best, best_p = None, -1
    for ch in product([0, 1], repeat=len(loads)):
        d = sum(c * l[1] for c, l in zip(ch, loads))
        p = sum(c * l[2] for c, l in zip(ch, loads))
        if d <= cap and p > best_p:
            best, best_p = ch, p
    return list(best), best_p

def greedy(loads, cap):
    # pick loads with the best priority-per-kW first
    order = sorted(range(len(loads)), key=lambda i: loads[i][2] / loads[i][1], reverse=True)
    ch, used = [0] * len(loads), 0
    for i in order:
        if used + loads[i][1] <= cap:
            ch[i] = 1
            used += loads[i][1]
    return ch, sum(c * l[2] for c, l in zip(ch, loads))

results = {}

t = time.time(); results["Brute force"] = (*brute_force(loads, battery_capacity), time.time() - t)
t = time.time(); results["Greedy"] = (*greedy(loads, battery_capacity), time.time() - t)
t = time.time()
r = solver.solve(qp)
results["QAOA"] = ([int(v) for v in r.x], r.fval, time.time() - t)

print(f"{'Method':12} {'Choice':18} {'Priority':>8} {'Time (s)':>9}")
for name, (choice, pri, sec) in results.items():
    print(f"{name:12} {str(choice):18} {pri:>8} {sec:>9.3f}")
```

```python
loads = [
    ("Hospital",      6, 7),
    ("Cell Tower",    5, 5),
    ("Water Pump",    5, 5),
    ("Shelter",       4, 1),
]
battery_capacity = 10

from qiskit_optimization import QuadraticProgram
qp = QuadraticProgram("Q-RESTORE-hard")
for i in range(len(loads)):
    qp.binary_var(name=f"x{i}")
qp.maximize(linear={f"x{i}": loads[i][2] for i in range(len(loads))})
qp.linear_constraint(
    linear={f"x{i}": loads[i][1] for i in range(len(loads))},
    sense="<=", rhs=battery_capacity, name="battery",
)

results = {}
t = time.time(); results["Brute force"] = (*brute_force(loads, battery_capacity), time.time() - t)
t = time.time(); results["Greedy"] = (*greedy(loads, battery_capacity), time.time() - t)

# QAOA: 3 runs to see how reliable it is
for k in range(3):
    t = time.time()
    r = solver.solve(qp)
    results[f"QAOA run {k+1}"] = ([int(v) for v in r.x], r.fval, time.time() - t)

print(f"{'Method':12} {'Choice':18} {'Priority':>8} {'Time (s)':>9}")
for name, (choice, pri, sec) in results.items():
    print(f"{name:12} {str(choice):18} {pri:>8} {sec:>9.3f}")
```

```python
import matplotlib.pyplot as plt

names = list(results.keys())
priorities = [float(results[n][1]) for n in names]
colors = ["#2e7d32" if n == "Brute force" else
          "#c62828" if n == "Greedy" else "#1565c0" for n in names]

plt.figure(figsize=(8, 4.5))
bars = plt.bar(names, priorities, color=colors)
plt.bar_label(bars, fmt="%.0f")
plt.ylabel("Total priority served")
plt.title("Q-RESTORE: Greedy vs. Brute Force vs. QAOA")
plt.xticks(rotation=20)
plt.tight_layout()
plt.savefig("q_restore_results.png", dpi=200)
plt.show()
```

```python
%%writefile app.py
import time
from itertools import product
import pandas as pd
import streamlit as st
from qiskit.primitives import StatevectorSampler
from qiskit_algorithms import QAOA
from qiskit_algorithms.optimizers import COBYLA
from qiskit_optimization import QuadraticProgram
from qiskit_optimization.algorithms import MinimumEigenOptimizer

st.title("⚡ Q-RESTORE")
st.caption("Quantum-assisted emergency power restoration")

capacity = st.number_input("Battery capacity (units)", 1, 100, 10)
df = st.data_editor(
    pd.DataFrame({
        "Load": ["Hospital", "Cell Tower", "Water Pump", "Shelter"],
        "Demand": [6, 5, 5, 4],
        "Priority": [7, 5, 5, 1],
    }),
    num_rows="dynamic",
)
loads = [(r.Load, int(r.Demand), int(r.Priority))
         for r in df.dropna().itertuples() if r.Demand > 0]

def brute_force(loads, cap):
    best, best_p = [0] * len(loads), 0
    for ch in product([0, 1], repeat=len(loads)):
        d = sum(c * l[1] for c, l in zip(ch, loads))
        p = sum(c * l[2] for c, l in zip(ch, loads))
        if d <= cap and p > best_p:
            best, best_p = list(ch), p
    return best, best_p

def greedy(loads, cap):
    order = sorted(range(len(loads)), key=lambda i: loads[i][2] / loads[i][1], reverse=True)
    ch, used = [0] * len(loads), 0
    for i in order:
        if used + loads[i][1] <= cap:
            ch[i], used = 1, used + loads[i][1]
    return ch, sum(c * l[2] for c, l in zip(ch, loads))

def run_qaoa(loads, cap):
    qp = QuadraticProgram("Q-RESTORE")
    for i in range(len(loads)):
        qp.binary_var(name=f"x{i}")
    qp.maximize(linear={f"x{i}": loads[i][2] for i in range(len(loads))})
    qp.linear_constraint(
        linear={f"x{i}": loads[i][1] for i in range(len(loads))},
        sense="<=", rhs=cap, name="battery")
    qaoa = QAOA(sampler=StatevectorSampler(), optimizer=COBYLA(maxiter=50), reps=1)
    r = MinimumEigenOptimizer(qaoa).solve(qp)
    return [int(v) for v in r.x], float(r.fval)

if st.button("Restore power"):
    if not loads:
        st.warning("Add at least one load.")
        st.stop()
    rows = []
    for name, fn in [("Brute force", brute_force), ("Greedy", greedy)]:
        t = time.time(); ch, p = fn(loads, capacity)
        rows.append((name, ch, p, round(time.time() - t, 3)))
    if len(loads) <= 6:
        with st.spinner("Running QAOA on the quantum simulator..."):
            t = time.time(); ch, p = run_qaoa(loads, capacity)
            rows.append(("QAOA", ch, p, round(time.time() - t, 1)))
    else:
        st.info("QAOA is limited to 6 loads in this demo (simulator speed).")

    res = pd.DataFrame(rows, columns=["Method", "Choice", "Priority", "Time (s)"])
    res["Choice"] = res["Choice"].astype(str)
    st.dataframe(res, hide_index=True)
    st.bar_chart(res.set_index("Method")["Priority"])
    best = rows[0][1]
    st.success("Powered loads (optimal): " + ", ".join(
        l[0] for c, l in zip(best, loads) if c == 1))
```
